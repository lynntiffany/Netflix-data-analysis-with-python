## Data Cleaning

```python
#Check for misssing values
df.isnull().sum()
```

```python
#Check for blank values
(df == " ").sum()
```

```python
# Check for leading or trailing spaces
for col in df.select_dtypes(include='object').columns:
    spaces = (df[col].dropna() != df[col].dropna().str.strip()).sum()
    print(col, spaces)
```

```python
df[df['title'].str.contains(r'\s{2,}', na=False)]['title']
```

```python
# Fix the whitespace
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
# Remove unnecessary multiple spaces within titles
df['title'] = df['title'].str.replace(r'\s+', ' ', regex=True)
```

```python
# Standardize Type
df['type'] = df['type'].str.title()
```

```python
df['type'].value_counts()
```

```python
#Check rating
df['rating'].value_counts()
```

```python
df['rating'] = df['rating'].str.strip().str.upper()
```

```python
df['rating'].value_counts()
```

```python
#Check country
df['country'] = df['country'].str.strip()
```

```python
df['country'].value_counts().head(20)
```

```python
#Check date data type
df['date_added'].dtype
```

```python
#Convert to datetime
df['date_added'] = pd.to_datetime(df['date_added'])
```

```python
df.info()
```

```python
#Handling invalid dates
df['date_added'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)
```

```python
# Checking for invalid dates created during conversion
df['date_added'].isna().sum()
```

#### Action taken
The `date_added` column was converted from string format to datetime format using pd.to_datetime() to ensure consistency and enable time-based analysis
